```python
# ======================================================================
# 🇱🇰 SRI LANKAN EMPLOYEE ATTRITION PREDICTION
# FINAL CLEAN RESEARCH + VIVA VERSION
#
# Dataset:
# Sri Lankan Employee Attrition & Workforce Experience Survey
#
# Dataset:
# 835 employees
# 29 original columns
#
# REAL TARGET:
# Employment Status / Turnover Event
#
# MODELS:
# 1. Logistic Regression
# 2. Random Forest
# 3. XGBoost
# 4. CatBoost
#
# PRIMARY MODEL SELECTION METRIC:
# PR-AUC
#
# FINAL VISUALIZATIONS:
# 1. Target Distribution
# 2. Model Comparison
# 3. Confusion Matrix
# 4. Feature Importance
#
# TEST SET:
# Used ONLY for final evaluation
#
# NO SYNTHETIC TARGET
# ======================================================================


# ======================================================================
# 1. INSTALL REQUIRED PACKAGES
# ======================================================================

!pip install -q catboost xgboost joblib


# ======================================================================
# 2. IMPORT LIBRARIES
# ======================================================================

import os
import json
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    RepeatedStratifiedKFold,
    RandomizedSearchCV,
    cross_val_predict
)

from sklearn.pipeline import Pipeline

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score,
    average_precision_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

from scipy.stats import (
    loguniform,
    randint,
    uniform
)

from xgboost import XGBClassifier

from catboost import CatBoostClassifier


warnings.filterwarnings("ignore")

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.width",
    200
)


# ======================================================================
# 3. LOAD DATASET
# ======================================================================

print("\n" + "=" * 80)
print("📂 LOAD DATASET")
print("=" * 80)

uploaded = files.upload()

filename = list(
    uploaded.keys()
)[0]

df = pd.read_csv(
    filename
)

print("\n✅ Dataset loaded")

print(
    "File:",
    filename
)

print(
    "Shape:",
    df.shape
)


# ======================================================================
# 4. DATASET OVERVIEW
# ======================================================================

print("\n" + "=" * 80)
print("📊 DATASET OVERVIEW")
print("=" * 80)

print(
    "Rows:",
    df.shape[0]
)

print(
    "Columns:",
    df.shape[1]
)

display(
    df.head()
)


# ======================================================================
# 5. DEFINE REAL TARGET
# ======================================================================

TARGET = "Employment Status / Turnover Event"


if TARGET not in df.columns:

    raise ValueError(
        f"Target column not found: {TARGET}"
    )


print("\n" + "=" * 80)
print("🎯 TARGET VARIABLE")
print("=" * 80)

print(
    "Target:",
    TARGET
)

print("\nOriginal target distribution:")

display(
    df[TARGET].value_counts(
        dropna=False
    )
)


# ======================================================================
# 6. TARGET ENCODING
# ======================================================================

target_clean = (

    df[TARGET]
    .astype(str)
    .str.strip()
    .str.lower()

)


target_mapping = {

    "no": 0,
    "yes": 1

}


y = target_clean.map(
    target_mapping
)


if y.isna().any():

    unexpected = (

        df.loc[
            y.isna(),
            TARGET
        ]
        .unique()

    )

    raise ValueError(
        f"Unexpected target values: {unexpected}"
    )


print("\n" + "=" * 80)
print("🔄 TARGET ENCODING")
print("=" * 80)

print(
    "0 = No Turnover"
)

print(
    "1 = Turnover"
)

print("\nEncoded target counts:")

display(
    y.value_counts()
)

print("\nTarget percentages:")

display(
    (
        y.value_counts(
            normalize=True
        )
        * 100
    ).round(2)
)


# ======================================================================
# 7. DATA QUALITY CHECK
# ======================================================================

print("\n" + "=" * 80)
print("🔍 DATA QUALITY CHECK")
print("=" * 80)


missing_table = pd.DataFrame({

    "Missing_Count":
        df.isna().sum(),

    "Missing_Percentage":
        (
            df.isna().mean()
            * 100
        ).round(2)

})


missing_table = (

    missing_table
    .sort_values(
        "Missing_Count",
        ascending=False
    )

)


display(
    missing_table
)


duplicate_count = (
    df.duplicated().sum()
)


print(
    "\nDuplicate rows:",
    duplicate_count
)


# ======================================================================
# 8. TARGET DISTRIBUTION DIAGRAM
# ======================================================================

plt.figure(
    figsize=(7, 5)
)

target_counts = (
    df[TARGET]
    .value_counts()
)


bars = plt.bar(

    target_counts.index.astype(str),

    target_counts.values

)


plt.title(
    "Employee Turnover Distribution",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    "Turnover Event"
)

plt.ylabel(
    "Number of Employees"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height,

        str(int(height)),

        ha="center",

        va="bottom"

    )


plt.tight_layout()

plt.savefig(
    "01_target_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ======================================================================
# 9. REMOVE UNNECESSARY / ADMINISTRATIVE COLUMNS
# ======================================================================

print("\n" + "=" * 80)
print("🚫 FEATURE EXCLUSION")
print("=" * 80)


# Timestamp is administrative metadata.
TIMESTAMP_COLUMNS = [

    c

    for c in df.columns

    if c.strip().lower()
    == "timestamp"

]


# Consent question is administrative
# and should not be a predictive feature.
CONSENT_COLUMNS = [

    c

    for c in df.columns

    if c.startswith(
        "I have read"
    )

]


EXCLUDED_COLUMNS = (

    TIMESTAMP_COLUMNS
    +
    CONSENT_COLUMNS
    +
    [TARGET]

)


EXCLUDED_COLUMNS = list(
    dict.fromkeys(
        EXCLUDED_COLUMNS
    )
)


print(
    "\nExcluded columns:"
)

for column in EXCLUDED_COLUMNS:

    print(
        " -",
        column
    )


# ======================================================================
# 10. CREATE FEATURES
# ======================================================================

X = (

    df
    .drop(
        columns=EXCLUDED_COLUMNS,
        errors="ignore"
    )
    .copy()

)


print("\n" + "=" * 80)
print("📐 FINAL FEATURE MATRIX")
print("=" * 80)

print(
    "Rows:",
    X.shape[0]
)

print(
    "Predictors:",
    X.shape[1]
)


# ======================================================================
# 11. IDENTIFY FEATURE TYPES
# ======================================================================

categorical_features = (

    X
    .select_dtypes(
        include=[
            "object",
            "category"
        ]
    )
    .columns
    .tolist()

)


numerical_features = [

    c

    for c in X.columns

    if c not in categorical_features

]


print(
    "\nCategorical features:",
    len(categorical_features)
)

print(
    "Numerical features:",
    len(numerical_features)
)


# ======================================================================
# 12. TRAIN / TEST SPLIT
# ======================================================================

print("\n" + "=" * 80)
print("✂️ STRATIFIED TRAIN / TEST SPLIT")
print("=" * 80)


X_train, X_test, y_train, y_test = (

    train_test_split(

        X,

        y,

        test_size=0.20,

        stratify=y,

        random_state=42

    )

)


print(
    "Training records:",
    len(X_train)
)

print(
    "Testing records:",
    len(X_test)
)


print("\nTraining distribution:")

display(
    y_train.value_counts()
)


print("\nTesting distribution:")

display(
    y_test.value_counts()
)


# ======================================================================
# 13. PREPROCESSING
# ======================================================================

numeric_pipeline = Pipeline([

    (
        "imputer",

        SimpleImputer(
            strategy="median"
        )

    ),

    (
        "scaler",

        StandardScaler()

    )

])


categorical_pipeline = Pipeline([

    (
        "imputer",

        SimpleImputer(
            strategy="most_frequent"
        )

    ),

    (
        "encoder",

        OneHotEncoder(
            handle_unknown="ignore"
        )

    )

])


preprocessor = ColumnTransformer([

    (
        "numeric",

        numeric_pipeline,

        numerical_features

    ),

    (
        "categorical",

        categorical_pipeline,

        categorical_features

    )

])


# ======================================================================
# 14. CROSS-VALIDATION
# ======================================================================

CV = RepeatedStratifiedKFold(

    n_splits=5,

    n_repeats=2,

    random_state=42

)


print("\n" + "=" * 80)
print("🔁 CROSS-VALIDATION")
print("=" * 80)

print(
    "5 folds × 2 repeats = 10 validation folds"
)

print(
    "Primary metric = PR-AUC"
)


# ======================================================================
# 15. LOGISTIC REGRESSION
# ======================================================================

print("\n" + "=" * 80)
print("1️⃣ LOGISTIC REGRESSION")
print("=" * 80)


lr_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",

        LogisticRegression(

            max_iter=3000,

            random_state=42

        )

    )

])


lr_search = RandomizedSearchCV(

    lr_pipeline,

    {

        "classifier__C":
            loguniform(
                0.01,
                10
            )

    },

    n_iter=20,

    scoring="average_precision",

    cv=CV,

    random_state=42,

    n_jobs=-1,

    refit=True

)


lr_search.fit(

    X_train,

    y_train

)


lr_model = (
    lr_search.best_estimator_
)


lr_cv_score = (
    lr_search.best_score_
)


print(
    "CV PR-AUC:",
    round(
        lr_cv_score,
        4
    )
)


# ======================================================================
# 16. RANDOM FOREST
# ======================================================================

print("\n" + "=" * 80)
print("2️⃣ RANDOM FOREST")
print("=" * 80)


rf_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",

        RandomForestClassifier(

            class_weight="balanced",

            random_state=42,

            n_jobs=-1

        )

    )

])


rf_search = RandomizedSearchCV(

    rf_pipeline,

    {

        "classifier__n_estimators":
            randint(
                300,
                800
            ),

        "classifier__max_depth":
            [
                None,
                4,
                6,
                8,
                12
            ],

        "classifier__min_samples_leaf":
            randint(
                1,
                8
            ),

        "classifier__max_features":
            [
                "sqrt",
                "log2",
                0.5
            ]

    },

    n_iter=25,

    scoring="average_precision",

    cv=CV,

    random_state=42,

    n_jobs=-1,

    refit=True

)


rf_search.fit(

    X_train,

    y_train

)


rf_model = (
    rf_search.best_estimator_
)


rf_cv_score = (
    rf_search.best_score_
)


print(
    "CV PR-AUC:",
    round(
        rf_cv_score,
        4
    )
)


# ======================================================================
# 17. XGBOOST
# ======================================================================

print("\n" + "=" * 80)
print("3️⃣ XGBOOST")
print("=" * 80)


xgb_pipeline = Pipeline([

    (
        "preprocessor",
        preprocessor
    ),

    (
        "classifier",

        XGBClassifier(

            objective="binary:logistic",

            eval_metric="logloss",

            tree_method="hist",

            random_state=42,

            n_jobs=-1,

            subsample=0.8,

            colsample_bytree=0.8

        )

    )

])


xgb_search = RandomizedSearchCV(

    xgb_pipeline,

    {

        "classifier__n_estimators":
            randint(
                200,
                800
            ),

        "classifier__max_depth":
            randint(
                2,
                7
            ),

        "classifier__learning_rate":
            loguniform(
                0.01,
                0.15
            ),

        "classifier__min_child_weight":
            randint(
                1,
                10
            ),

        "classifier__gamma":
            uniform(
                0,
                2
            ),

        "classifier__reg_lambda":
            loguniform(
                0.5,
                20
            ),

        "classifier__reg_alpha":
            loguniform(
                0.001,
                2
            )

    },

    n_iter=25,

    scoring="average_precision",

    cv=CV,

    random_state=42,

    n_jobs=-1,

    refit=True

)


xgb_search.fit(

    X_train,

    y_train

)


xgb_model = (
    xgb_search.best_estimator_
)


xgb_cv_score = (
    xgb_search.best_score_
)


print(
    "CV PR-AUC:",
    round(
        xgb_cv_score,
        4
    )
)


# ======================================================================
# 18. CATBOOST
# ======================================================================

print("\n" + "=" * 80)
print("4️⃣ CATBOOST")
print("=" * 80)


X_train_cb = X_train.copy()

X_test_cb = X_test.copy()


for column in categorical_features:

    X_train_cb[column] = (

        X_train_cb[column]
        .fillna("Missing")
        .astype(str)

    )

    X_test_cb[column] = (

        X_test_cb[column]
        .fillna("Missing")
        .astype(str)

    )


cat_configs = [

    {
        "iterations": 300,
        "depth": 3,
        "learning_rate": 0.03,
        "l2_leaf_reg": 10
    },

    {
        "iterations": 400,
        "depth": 4,
        "learning_rate": 0.03,
        "l2_leaf_reg": 10
    },

    {
        "iterations": 500,
        "depth": 4,
        "learning_rate": 0.03,
        "l2_leaf_reg": 15
    },

    {
        "iterations": 600,
        "depth": 5,
        "learning_rate": 0.03,
        "l2_leaf_reg": 15
    },

    {
        "iterations": 700,
        "depth": 4,
        "learning_rate": 0.02,
        "l2_leaf_reg": 15
    }

]


cat_cv_results = []


for config in cat_configs:

    fold_scores = []


    for train_idx, valid_idx in CV.split(

        X_train_cb,

        y_train

    ):

        Xtr = (
            X_train_cb
            .iloc[train_idx]
            .copy()
        )

        Xval = (
            X_train_cb
            .iloc[valid_idx]
            .copy()
        )

        ytr = (
            y_train
            .iloc[train_idx]
        )

        yval = (
            y_train
            .iloc[valid_idx]
        )


        model = CatBoostClassifier(

            **config,

            loss_function="Logloss",

            verbose=False,

            random_seed=42,

            thread_count=-1

        )


        model.fit(

            Xtr,

            ytr,

            cat_features=categorical_features

        )


        probability = (

            model
            .predict_proba(
                Xval
            )[:, 1]

        )


        score = (

            average_precision_score(

                yval,

                probability

            )

        )


        fold_scores.append(
            score
        )


    cat_cv_results.append({

        "Configuration":
            str(config),

        "Mean_PR_AUC":
            np.mean(
                fold_scores
            ),

        "Std_PR_AUC":
            np.std(
                fold_scores
            )

    })


cat_cv_results = (

    pd.DataFrame(
        cat_cv_results
    )
    .sort_values(
        "Mean_PR_AUC",
        ascending=False
    )
    .reset_index(
        drop=True
    )

)


best_cat_index = 0


best_cat_config = eval(

    cat_cv_results
    .iloc[
        best_cat_index
    ]["Configuration"]

)


cat_cv_score = (

    cat_cv_results
    .iloc[
        best_cat_index
    ]["Mean_PR_AUC"]

)


cat_cv_std = (

    cat_cv_results
    .iloc[
        best_cat_index
    ]["Std_PR_AUC"]

)


print(
    "Best CatBoost CV PR-AUC:",
    round(
        cat_cv_score,
        4
    )
)

print(
    "CV Std:",
    round(
        cat_cv_std,
        4
    )
)

print(
    "Best configuration:",
    best_cat_config
)


# ======================================================================
# 19. MODEL COMPARISON
# ======================================================================

print("\n" + "=" * 80)
print("🏆 MODEL COMPARISON")
print("=" * 80)


cv_results = pd.DataFrame({

    "Model": [

        "Logistic Regression",

        "Random Forest",

        "XGBoost",

        "CatBoost"

    ],

    "CV_PR_AUC": [

        lr_cv_score,

        rf_cv_score,

        xgb_cv_score,

        cat_cv_score

    ]

})


cv_results = (

    cv_results
    .sort_values(
        "CV_PR_AUC",
        ascending=False
    )
    .reset_index(
        drop=True
    )

)


display(
    cv_results.round(4)
)


BEST_MODEL_NAME = (

    cv_results
    .iloc[0]["Model"]

)


print(
    "\n🥇 Selected model:",
    BEST_MODEL_NAME
)

print(
    "Selection criterion: highest repeated-CV PR-AUC"
)


# ======================================================================
# 20. MODEL COMPARISON DIAGRAM
# ======================================================================

plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(

    cv_results["Model"],

    cv_results["CV_PR_AUC"]

)


plt.title(
    "Model Comparison Using Repeated CV PR-AUC",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Mean Cross-Validated PR-AUC"
)


plt.xticks(
    rotation=20,
    ha="right"
)


for bar in bars:

    height = bar.get_height()

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height,

        f"{height:.3f}",

        ha="center",

        va="bottom"

    )


plt.tight_layout()


plt.savefig(

    "02_model_comparison.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# 21. OOF THRESHOLD OPTIMIZATION
# ======================================================================

print("\n" + "=" * 80)
print("🎯 TRAINING-ONLY THRESHOLD OPTIMIZATION")
print("=" * 80)


def optimize_threshold(

    y_true,

    probabilities

):

    thresholds = np.arange(

        0.05,

        0.901,

        0.005

    )


    results = []


    for threshold in thresholds:

        predictions = (

            probabilities
            >= threshold

        ).astype(int)


        results.append({

            "Threshold":
                threshold,

            "F1":
                f1_score(

                    y_true,

                    predictions,

                    zero_division=0

                ),

            "Precision":
                precision_score(

                    y_true,

                    predictions,

                    zero_division=0

                ),

            "Recall":
                recall_score(

                    y_true,

                    predictions,

                    zero_division=0

                ),

            "BalancedAccuracy":
                balanced_accuracy_score(

                    y_true,

                    predictions

                )

        })


    results_df = pd.DataFrame(
        results
    )


    best_row = (

        results_df
        .loc[
            results_df["F1"].idxmax()
        ]

    )


    return (

        results_df,

        float(
            best_row["Threshold"]
        )

    )


# ----------------------------------------------------------------------
# Pipeline models
# ----------------------------------------------------------------------

pipeline_models = {

    "Logistic Regression":
        lr_model,

    "Random Forest":
        rf_model,

    "XGBoost":
        xgb_model

}


threshold_results = {}


OOF_CV = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42

)


for name, model in pipeline_models.items():

    print(
        "\nOOF prediction:",
        name
    )


    oof_probability = cross_val_predict(

        model,

        X_train,

        y_train,

        cv=OOF_CV,

        method="predict_proba",

        n_jobs=-1

    )[:, 1]


    threshold_df_temp, threshold = (

        optimize_threshold(

            y_train,

            oof_probability

        )

    )


    best_row = (

        threshold_df_temp
        .loc[
            threshold_df_temp["F1"].idxmax()
        ]

    )


    threshold_results[name] = {

        "Threshold":
            threshold,

        "F1":
            float(
                best_row["F1"]
            ),

        "Precision":
            float(
                best_row["Precision"]
            ),

        "Recall":
            float(
                best_row["Recall"]
            )

    }


# ----------------------------------------------------------------------
# CatBoost OOF
# ----------------------------------------------------------------------

print(
    "\nOOF prediction: CatBoost"
)


cat_oof_probability = np.zeros(

    len(X_train_cb)

)


for train_idx, valid_idx in OOF_CV.split(

    X_train_cb,

    y_train

):

    Xtr = (
        X_train_cb
        .iloc[train_idx]
        .copy()
    )

    Xval = (
        X_train_cb
        .iloc[valid_idx]
        .copy()
    )

    ytr = (
        y_train
        .iloc[train_idx]
    )


    model = CatBoostClassifier(

        **best_cat_config,

        loss_function="Logloss",

        verbose=False,

        random_seed=42,

        thread_count=-1

    )


    model.fit(

        Xtr,

        ytr,

        cat_features=categorical_features

    )


    cat_oof_probability[
        valid_idx
    ] = (

        model
        .predict_proba(
            Xval
        )[:, 1]

    )


cat_threshold_df, cat_threshold = (

    optimize_threshold(

        y_train,

        cat_oof_probability

    )

)


cat_best_row = (

    cat_threshold_df
    .loc[
        cat_threshold_df["F1"].idxmax()
    ]

)


threshold_results["CatBoost"] = {

    "Threshold":
        cat_threshold,

    "F1":
        float(
            cat_best_row["F1"]
        ),

    "Precision":
        float(
            cat_best_row["Precision"]
        ),

    "Recall":
        float(
            cat_best_row["Recall"]
        )

}


BEST_THRESHOLD = (

    threshold_results[
        BEST_MODEL_NAME
    ]["Threshold"]

)


print("\nSelected threshold:")

print(
    round(
        BEST_THRESHOLD,
        3
    )
)

print(
    "\nImportant: threshold was selected from training OOF predictions only."
)


# ======================================================================
# 22. TRAIN FINAL MODEL
# ======================================================================

print("\n" + "=" * 80)
print("🚀 TRAIN FINAL MODEL")
print("=" * 80)


if BEST_MODEL_NAME == "CatBoost":

    FINAL_MODEL = CatBoostClassifier(

        **best_cat_config,

        loss_function="Logloss",

        verbose=False,

        random_seed=42,

        thread_count=-1

    )


    FINAL_MODEL.fit(

        X_train_cb,

        y_train,

        cat_features=categorical_features

    )


    test_probability = (

        FINAL_MODEL
        .predict_proba(
            X_test_cb
        )[:, 1]

    )


else:

    FINAL_MODEL = pipeline_models[
        BEST_MODEL_NAME
    ]


    FINAL_MODEL.fit(

        X_train,

        y_train

    )


    test_probability = (

        FINAL_MODEL
        .predict_proba(
            X_test
        )[:, 1]

    )


print(
    "✅ Final model trained:",
    BEST_MODEL_NAME
)


# ======================================================================
# 23. FINAL TEST PREDICTIONS
# ======================================================================

y_pred = (

    test_probability
    >= BEST_THRESHOLD

).astype(int)


# ======================================================================
# 24. FINAL TEST METRICS
# ======================================================================

accuracy = accuracy_score(

    y_test,

    y_pred

)


precision = precision_score(

    y_test,

    y_pred,

    zero_division=0

)


recall = recall_score(

    y_test,

    y_pred,

    zero_division=0

)


f1 = f1_score(

    y_test,

    y_pred,

    zero_division=0

)


balanced_accuracy = (

    balanced_accuracy_score(

        y_test,

        y_pred

    )

)


roc_auc = roc_auc_score(

    y_test,

    test_probability

)


pr_auc = average_precision_score(

    y_test,

    test_probability

)


mcc = matthews_corrcoef(

    y_test,

    y_pred

)


cm = confusion_matrix(

    y_test,

    y_pred

)


tn, fp, fn, tp = cm.ravel()


specificity = (

    tn /
    (tn + fp)

    if (tn + fp) > 0

    else 0

)


# ======================================================================
# 25. FINAL RESULTS
# ======================================================================

print("\n" + "=" * 80)
print("🇱🇰 FINAL TEST RESULTS")
print("=" * 80)


print(
    f"\nFinal Model        : {BEST_MODEL_NAME}"
)

print(
    f"Threshold          : {BEST_THRESHOLD:.3f}"
)

print(
    f"\nAccuracy           : {accuracy:.4f}"
)

print(
    f"Precision          : {precision:.4f}"
)

print(
    f"Recall             : {recall:.4f}"
)

print(
    f"F1 Score           : {f1:.4f}"
)

print(
    f"Balanced Accuracy  : {balanced_accuracy:.4f}"
)

print(
    f"ROC-AUC            : {roc_auc:.4f}"
)

print(
    f"PR-AUC             : {pr_auc:.4f}"
)

print(
    f"MCC                : {mcc:.4f}"
)

print(
    f"Specificity        : {specificity:.4f}"
)


# ======================================================================
# 26. CLASSIFICATION REPORT
# ======================================================================

print("\n" + "=" * 80)
print("📋 CLASSIFICATION REPORT")
print("=" * 80)


print(

    classification_report(

        y_test,

        y_pred,

        target_names=[

            "No Turnover",

            "Turnover"

        ],

        digits=4

    )

)


# ======================================================================
# 27. CONFUSION MATRIX
# ======================================================================

print("\n" + "=" * 80)
print("📊 CONFUSION MATRIX")
print("=" * 80)


print(cm)


print(
    "\nTrue Negative :",
    tn
)

print(
    "False Positive:",
    fp
)

print(
    "False Negative:",
    fn
)

print(
    "True Positive  :",
    tp
)


# ======================================================================
# 28. CONFUSION MATRIX DIAGRAM
# ======================================================================

plt.figure(
    figsize=(7, 6)
)


plt.imshow(cm)


plt.title(
    "Confusion Matrix - Final Attrition Model",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)


plt.xticks(

    [0, 1],

    [
        "No Turnover",
        "Turnover"
    ]

)


plt.yticks(

    [0, 1],

    [
        "No Turnover",
        "Turnover"
    ]

)


for i in range(2):

    for j in range(2):

        plt.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            fontsize=16,

            fontweight="bold"

        )


plt.colorbar()

plt.tight_layout()


plt.savefig(

    "03_confusion_matrix.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# 29. FEATURE IMPORTANCE
# ======================================================================

print("\n" + "=" * 80)
print("🔎 FEATURE IMPORTANCE")
print("=" * 80)


if BEST_MODEL_NAME == "CatBoost":

    importance_values = (

        FINAL_MODEL
        .get_feature_importance()

    )


    feature_importance = pd.DataFrame({

        "Feature":
            X_train.columns,

        "Importance":
            importance_values

    })


else:

    fitted_preprocessor = (

        FINAL_MODEL
        .named_steps[
            "preprocessor"
        ]

    )


    transformed_names = (

        fitted_preprocessor
        .get_feature_names_out()

    )


    classifier = (

        FINAL_MODEL
        .named_steps[
            "classifier"
        ]

    )


    if hasattr(

        classifier,

        "feature_importances_"

    ):

        importance_values = (

            classifier
            .feature_importances_

        )


        feature_importance = pd.DataFrame({

            "Feature":
                transformed_names,

            "Importance":
                importance_values

        })


    else:

        coefficients = (

            classifier
            .coef_[0]

        )


        feature_importance = pd.DataFrame({

            "Feature":
                transformed_names,

            "Coefficient":
                coefficients,

            "AbsoluteImportance":
                np.abs(
                    coefficients
                )

        })


        feature_importance = (

            feature_importance
            .sort_values(
                "AbsoluteImportance",
                ascending=False
            )
            .reset_index(
                drop=True
            )

        )


if "Importance" in feature_importance.columns:

    feature_importance = (

        feature_importance
        .sort_values(
            "Importance",
            ascending=False
        )
        .reset_index(
            drop=True
        )

    )


display(
    feature_importance.head(15)
)


# ======================================================================
# 30. FEATURE IMPORTANCE DIAGRAM
# ======================================================================

if "Importance" in feature_importance.columns:

    top_features = (

        feature_importance
        .head(15)
        .sort_values(
            "Importance"
        )

    )

    values = (
        top_features[
            "Importance"
        ]
    )

    xlabel = (
        "Feature Importance"
    )

else:

    top_features = (

        feature_importance
        .head(15)
        .sort_values(
            "AbsoluteImportance"
        )

    )

    values = (

        top_features[
            "AbsoluteImportance"
        ]

    )

    xlabel = (
        "Absolute Coefficient"
    )


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    top_features["Feature"],

    values

)


plt.xlabel(
    xlabel
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 15 Features Associated With Attrition Prediction",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()


plt.savefig(

    "04_feature_importance.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# 31. FINAL PREDICTION DATASET
# ======================================================================

predictions = X_test.copy()


predictions[
    "ActualTurnover"
] = y_test.values


predictions[
    "AttritionProbability"
] = (

    test_probability * 100

).round(2)


predictions[
    "PredictedTurnover"
] = y_pred

# ======================================================================
# 31A. ADD RETENTION RISK DRIVER VALUES
# ======================================================================

# LowMatchRisk is intentionally left unavailable because this attrition
# dataset does not contain CV-to-job matching information.
predictions["LowMatchRisk"] = np.nan


# CompanySwitchRate is a tenure-normalized switching proxy.
previous_companies = pd.to_numeric(
    predictions[
        "Total Number of Previous Companies Worked For"
    ],
    errors="coerce"
)

current_company_years = pd.to_numeric(
    predictions[
        "Total Years of Service with Current Company"
    ],
    errors="coerce"
)

predictions["CompanySwitchRate"] = (
    previous_companies
    /
    current_company_years.clip(lower=1)
)


# SalaryGapRisk measures the proportional gap when expected salary
# exceeds current salary.
current_salary = pd.to_numeric(
    predictions[
        "Current Monthly Base Salary (LKR)"
    ],
    errors="coerce"
)

expected_salary = pd.to_numeric(
    predictions[
        "Expected Monthly Salary for New Role / Progression (LKR)"
    ],
    errors="coerce"
)

predictions["SalaryGapRisk"] = (
    (
        expected_salary
        -
        current_salary
    )
    /
    expected_salary.replace(0, np.nan)
).clip(lower=0)


print("\n" + "=" * 80)
print("🧩 RETENTION RISK DRIVERS")
print("=" * 80)

display(
    predictions[
        [
            "LowMatchRisk",
            "CompanySwitchRate",
            "SalaryGapRisk"
        ]
    ].head(20)
)


# ======================================================================
# 32. SIMPLE RISK CATEGORY
# ======================================================================

predictions[
    "RiskLevel"
] = pd.cut(

    predictions[
        "AttritionProbability"
    ],

    bins=[

        -1,

        30,

        60,

        100

    ],

    labels=[

        "Low",

        "Medium",

        "High"

    ]

)


print("\n" + "=" * 80)
print("📊 FINAL PREDICTION SAMPLE")
print("=" * 80)


display(
    predictions.head(10)
)


# ======================================================================
# 33. FINAL METRICS TABLE
# ======================================================================

final_metrics = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "Balanced Accuracy",

        "ROC-AUC",

        "PR-AUC",

        "MCC",

        "Specificity",

        "Decision Threshold"

    ],

    "Value": [

        accuracy,

        precision,

        recall,

        f1,

        balanced_accuracy,

        roc_auc,

        pr_auc,

        mcc,

        specificity,

        BEST_THRESHOLD

    ]

})


print("\n" + "=" * 80)
print("📈 FINAL METRICS")
print("=" * 80)


display(
    final_metrics.round(4)
)


# ======================================================================
# 34. SAVE MODEL
# ======================================================================

model_package = {

    "model":
        FINAL_MODEL,

    "model_type":
        BEST_MODEL_NAME,

    "target":
        TARGET,

    "target_encoding": {

        "No": 0,

        "Yes": 1

    },

    "decision_threshold":
        float(
            BEST_THRESHOLD
        ),

    "selection_metric":
        "PR-AUC",

    "threshold_selection":
        "Training-only OOF F1 optimization",

    "cross_validation":
        "Repeated Stratified 5-Fold CV, 2 repeats",

    "features":
        list(
            X.columns
        ),

    "categorical_features":
        categorical_features,

    "numerical_features":
        numerical_features,

    "excluded_columns":
        EXCLUDED_COLUMNS,

    "test_metrics": {

        "accuracy":
            float(
                accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1":
            float(
                f1
            ),

        "balanced_accuracy":
            float(
                balanced_accuracy
            ),

        "roc_auc":
            float(
                roc_auc
            ),

        "pr_auc":
            float(
                pr_auc
            ),

        "mcc":
            float(
                mcc
            ),

        "specificity":
            float(
                specificity
            )

    }

}


joblib.dump(

    model_package,

    "Final_Sri_Lankan_Attrition_Model.pkl"

)


# ======================================================================
# 35. SAVE METADATA
# ======================================================================

metadata = {

    key: value

    for key, value
    in model_package.items()

    if key != "model"

}


with open(

    "model_metadata.json",

    "w"

) as f:

    json.dump(

        metadata,

        f,

        indent=2

    )


# ======================================================================
# 36. SAVE RESULTS
# ======================================================================

final_metrics.to_csv(

    "Final_Model_Metrics.csv",

    index=False

)


cv_results.to_csv(

    "Model_Selection_CV.csv",

    index=False

)


feature_importance.to_csv(

    "Feature_Importance.csv",

    index=False

)


predictions.to_csv(

    "Final_Test_Predictions.csv",

    index=False

)


# ======================================================================
# 37. SAVE RESEARCH SUMMARY
# ======================================================================

research_summary = {

    "Dataset": (
        "Sri Lankan Employee Attrition "
        "& Workforce Experience Survey"
    ),

    "Rows":
        int(
            df.shape[0]
        ),

    "Original_Columns":
        int(
            df.shape[1]
        ),

    "Predictors":
        int(
            X.shape[1]
        ),

    "Target":
        TARGET,

    "No_Turnover":
        int(
            (y == 0).sum()
        ),

    "Turnover":
        int(
            (y == 1).sum()
        ),

    "Final_Model":
        BEST_MODEL_NAME,

    "Model_Selection":
        "Highest repeated-CV PR-AUC",

    "Cross_Validation":
        "Repeated Stratified 5-Fold CV, 2 repeats",

    "Train_Test_Split":
        "80/20 stratified",

    "Threshold_Selection":
        "Training-only OOF F1 optimization",

    "Test_Set":
        "Untouched until final evaluation",

    "Accuracy":
        float(
            accuracy
        ),

    "Precision":
        float(
            precision
        ),

    "Recall":
        float(
            recall
        ),

    "F1":
        float(
            f1
        ),

    "Balanced_Accuracy":
        float(
            balanced_accuracy
        ),

    "ROC_AUC":
        float(
            roc_auc
        ),

    "PR_AUC":
        float(
            pr_auc
        ),

    "MCC":
        float(
            mcc
        )

}


with open(

    "Research_Model_Summary.json",

    "w"

) as f:

    json.dump(

        research_summary,

        f,

        indent=2

    )
# ======================================================================
# 38. FINAL OUTPUT FILES + DOWNLOAD
# ======================================================================

print("\n" + "=" * 80)
print("📁 FINAL RESEARCH OUTPUT FILES")
print("=" * 80)


output_files = [

    "Final_Sri_Lankan_Attrition_Model.pkl",

    "model_metadata.json",

    "Research_Model_Summary.json",

    "Final_Model_Metrics.csv",

    "Model_Selection_CV.csv",

    "Feature_Importance.csv",

    "Final_Test_Predictions.csv",

    "01_target_distribution.png",

    "02_model_comparison.png",

    "03_confusion_matrix.png",

    "04_feature_importance.png"

]


# ----------------------------------------------------------------------
# CHECK FILES
# ----------------------------------------------------------------------

existing_files = []
missing_files = []


for file_name in output_files:

    if os.path.exists(file_name):

        existing_files.append(file_name)

        print(
            "✓ READY:",
            file_name
        )

    else:

        missing_files.append(file_name)

        print(
            "✗ MISSING:",
            file_name
        )


# ----------------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------------

print("\n" + "-" * 80)

print(
    f"Files ready for download: {len(existing_files)}"
)

print(
    f"Files missing: {len(missing_files)}"
)


# ----------------------------------------------------------------------
# DOWNLOAD FILES
# ----------------------------------------------------------------------

if len(existing_files) > 0:

    print("\n" + "=" * 80)
    print("⬇️ DOWNLOADING FINAL RESEARCH FILES")
    print("=" * 80)


    for file_name in existing_files:

        print(
            "Downloading:",
            file_name
        )

        files.download(
            file_name
        )


    print("\n" + "=" * 80)
    print("✅ ALL AVAILABLE FILES SENT FOR DOWNLOAD")
    print("=" * 80)


else:

    print(
        "\n❌ No output files were found."
    )


# ----------------------------------------------------------------------
# MISSING FILES
# ----------------------------------------------------------------------

if len(missing_files) > 0:

    print("\n⚠️ Missing files:")

    for file_name in missing_files:

        print(
            " -",
            file_name
        )


print(
    "\nDataset:",
    "Sri Lankan Employee Attrition & Workforce Experience Survey"
)

print(
    "Observations:",
    df.shape[0]
)

print(
    "Original Variables:",
    df.shape[1]
)

print(
    "Predictor Variables:",
    X.shape[1]
)

print(
    "Actual Target:",
    TARGET
)

print(
    "Final Model:",
    BEST_MODEL_NAME
)

print(
    "Model Selection Metric:",
    "PR-AUC"
)

print(
    "Cross-Validation:",
    "Repeated Stratified 5-Fold CV × 2"
)

print(
    "Train/Test Split:",
    "80/20 Stratified"
)

print(
    "Decision Threshold:",
    round(
        BEST_THRESHOLD,
        3
    )
)

print("\nFinal Test Performance:")

print(
    f"Accuracy          : {accuracy:.4f}"
)

print(
    f"Precision         : {precision:.4f}"
)

print(
    f"Recall            : {recall:.4f}"
)

print(
    f"F1 Score          : {f1:.4f}"
)

print(
    f"Balanced Accuracy : {balanced_accuracy:.4f}"
)

print(
    f"ROC-AUC           : {roc_auc:.4f}"
)

print(
    f"PR-AUC            : {pr_auc:.4f}"
)

print(
    f"MCC               : {mcc:.4f}"
)


```
