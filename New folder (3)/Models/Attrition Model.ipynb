```python
# ================================================================
# 🚀 ATTRITION MODEL V10
# SRI LANKAN EMPLOYEE ATTRITION & WORKFORCE EXPERIENCE SURVEY
#
# HYBRID STACKING:
#   1. CatBoost
#   2. Logistic Regression
#   3. Logistic Regression Meta Learner
#
# NEW DATASET ONLY
# GOOGLE COLAB - SINGLE CELL
# ================================================================


# ================================================================
# 0. INSTALLATION
# ================================================================

!pip -q install catboost optuna


# ================================================================
# 1. IMPORTS
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import gc
import os
import joblib
import optuna
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.base import clone

from google.colab import files

from catboost import CatBoostClassifier

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold
)

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)


optuna.logging.set_verbosity(
    optuna.logging.WARNING
)


# ================================================================
# 2. GLOBAL SETTINGS
# ================================================================

RANDOM_STATE = 42

TEST_SIZE = 0.20

N_FOLDS = 5

N_TRIALS = 25

MIN_ATTRITION_RECALL = 0.50


# ================================================================
# 3. HEADER
# ================================================================

print("=" * 95)
print("🚀 ATTRITION MODEL V10 - HYBRID STACKING")
print("=" * 95)

print(
    "\n🇱🇰 Sri Lankan Employee Attrition & Workforce Experience Survey"
)

print(
    "\nModel Architecture:"
)

print(
    "  • CatBoost Base Learner"
)

print(
    "  • Logistic Regression Base Learner"
)

print(
    "  • Logistic Regression Meta Learner"
)


# ================================================================
# 4. UPLOAD NEW DATASET
# ================================================================

print("\n" + "=" * 95)
print("📤 UPLOAD NEW DATASET")
print("=" * 95)

print(
    "\nPlease upload the NEW Google Forms CSV dataset."
)

uploaded = files.upload()

if len(uploaded) == 0:
    raise ValueError(
        "❌ No file was uploaded."
    )

file_name = next(iter(uploaded))

if not file_name.lower().endswith(".csv"):

    raise ValueError(
        "❌ This model expects a CSV file."
    )


# ================================================================
# 5. LOAD DATASET
# ================================================================

df = pd.read_csv(
    file_name
)


print("\n✅ DATASET LOADED")
print("-" * 95)

print(
    "File    :",
    file_name
)

print(
    "Rows    :",
    len(df)
)

print(
    "Columns :",
    len(df.columns)
)

print(
    "Missing :",
    int(df.isna().sum().sum())
)


# ================================================================
# 6. ORIGINAL DATASET HEADERS
# ================================================================

print("\n" + "=" * 95)
print("📋 ORIGINAL DATASET TABLE HEADERS")
print("=" * 95)

for i, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i:02d}. {column}"
    )

print(
    "\nTotal Headers:",
    len(df.columns)
)


# ================================================================
# 7. CHECK DATASET SIZE
# ================================================================

if len(df) < 100:

    print(
        "\n⚠️ WARNING: Dataset contains fewer than 100 responses."
    )

else:

    print(
        "\n✅ Dataset size is sufficient for this modelling experiment."
    )


# ================================================================
# 8. RENAME SURVEY VARIABLES
# ================================================================

RENAME_MAP = {

    "Marital Status":
        "MaritalStatus",

    "Highest Level of Completed Education":
        "Education",

    "Academic Discipline / Field of Study":
        "FieldOfStudy",

    "Daily Commute Distance from Home to Workplace":
        "CommuteDistance",

    "Primary / Preferred Employment District in Sri Lanka":
        "District",

    "Functional Department":
        "Department",

    "Job Role / Professional Designation":
        "JobRole",

    "Seniority Level / Job Tier":
        "JobLevel",

    "Total Number of Previous Companies Worked For":
        "NumCompaniesWorked",

    "Total Years of Service with Current Company":
        "YearsAtCompany",

    "Years in Current Role":
        "YearsInCurrentRole",

    "Years Working with Current Manager":
        "YearsWithCurrManager",

    "Years Since Last Promotion":
        "YearsSinceLastPromotion",

    "Contractual Notice Period":
        "NoticePeriod",

    "Work Operating Model":
        "WorkModel",

    "Business Travel Frequency":
        "BusinessTravel",

    "Overtime Requirement":
        "OverTime",

    "Physical & Virtual Work Environment Satisfaction":
        "EnvironmentSatisfaction",

    "Overall Job & Role Satisfaction":
        "JobSatisfaction",

    "Interpersonal & Management Relationship Satisfaction":
        "RelationshipSatisfaction",

    "Work-Life Balance Rating":
        "WorkLifeBalance",

    "Job Involvement & Sense of Purpose":
        "JobInvolvement",

    "Current Monthly Base Salary (LKR)":
        "CurrentSalaryLKR",

    "Expected Monthly Salary for New Role / Progression (LKR)":
        "ExpectedSalaryLKR",

    "Employment Status / Turnover Event":
        "TurnoverEvent"
}


df = df.rename(
    columns=RENAME_MAP
)


# ================================================================
# 9. TARGET CREATION
# ================================================================

print("\n" + "=" * 95)
print("🎯 TARGET VARIABLE")
print("=" * 95)

TARGET_SOURCE = "TurnoverEvent"

if TARGET_SOURCE not in df.columns:

    raise ValueError(
        "❌ Target column 'Employment Status / Turnover Event' "
        "was not found."
    )


print(
    "\nOriginal target values:"
)

print(
    df[TARGET_SOURCE]
    .value_counts(dropna=False)
)


# Convert Yes / No → 1 / 0

df["Attrition"] = (

    df[TARGET_SOURCE]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({

        "no": 0,

        "yes": 1

    })
)


# ================================================================
# 10. VALIDATE TARGET
# ================================================================

if df["Attrition"].isna().any():

    unknown_values = (

        df.loc[
            df["Attrition"].isna(),
            TARGET_SOURCE
        ]
        .value_counts(dropna=False)
    )

    print(
        "\n❌ Unknown target values:"
    )

    print(
        unknown_values
    )

    raise ValueError(
        "\nUnknown values exist in the target column."
    )


TARGET = "Attrition"

y = df[TARGET].astype(int)


# ================================================================
# 11. TARGET DISTRIBUTION
# ================================================================

print("\n" + "=" * 95)
print("📊 TARGET DISTRIBUTION")
print("=" * 95)

target_counts = (
    y.value_counts()
    .sort_index()
)

for cls, count in target_counts.items():

    label = (
        "No Attrition"
        if cls == 0
        else
        "Attrition"
    )

    percentage = (
        count / len(y)
    ) * 100

    print(
        f"{label:18s}: "
        f"{count:4d} "
        f"({percentage:.2f}%)"
    )


majority_baseline = (
    y.value_counts(
        normalize=True
    ).max()
)


print(
    "\nMajority Baseline Accuracy:",
    f"{majority_baseline * 100:.2f}%"
)


# ================================================================
# 12. TARGET DISTRIBUTION VISUALIZATION
# ================================================================

plt.figure(figsize=(7, 5))

plt.bar(
    ["No Attrition", "Attrition"],
    [
        int((y == 0).sum()),
        int((y == 1).sum())
    ]
)

plt.title(
    "Attrition Class Distribution"
)

plt.ylabel(
    "Number of Employees"
)

plt.xlabel(
    "Class"
)

plt.tight_layout()

plt.show()


# ================================================================
# 13. REMOVE NON-PREDICTIVE / EXCLUDED VARIABLES
# ================================================================

# Consent question
consent_columns = [

    column

    for column in df.columns

    if (
        "agree to participate"
        in column.lower()
    )
]


DROP_COLUMNS = [

    # Target
    TARGET,

    TARGET_SOURCE,

    # Response timestamp
    "Timestamp",

    # Demographic variables intentionally excluded
    "Age",
    "Gender",
    "MaritalStatus"

]


DROP_COLUMNS += consent_columns


# Remove only columns that actually exist

DROP_COLUMNS = [

    column

    for column in DROP_COLUMNS

    if column in df.columns
]


print("\n" + "=" * 95)
print("🗑️ EXCLUDED COLUMNS")
print("=" * 95)

for column in DROP_COLUMNS:

    print(
        "•",
        column
    )


# ================================================================
# 14. CREATE MODEL INPUT
# ================================================================

FEATURES = [

    column

    for column in df.columns

    if column not in DROP_COLUMNS
]


X = df[
    FEATURES
].copy()


print("\n" + "=" * 95)
print("🧩 INITIAL FEATURE SET")
print("=" * 95)

print(
    "Initial Features:",
    len(FEATURES)
)


# ================================================================
# 15. CONVERT KNOWN NUMERIC VARIABLES
# ================================================================

NUMERIC_CANDIDATES = [

    "NumCompaniesWorked",

    "YearsAtCompany",

    "YearsInCurrentRole",

    "YearsWithCurrManager",

    "YearsSinceLastPromotion",

    "EnvironmentSatisfaction",

    "JobSatisfaction",

    "RelationshipSatisfaction",

    "WorkLifeBalance",

    "JobInvolvement",

    "CurrentSalaryLKR",

    "ExpectedSalaryLKR"

]


for column in NUMERIC_CANDIDATES:

    if column in X.columns:

        X[column] = pd.to_numeric(
            X[column],
            errors="coerce"
        )


# ================================================================
# 16. FEATURE ENGINEERING
# ================================================================

print("\n" + "=" * 95)
print("⚙️ FEATURE ENGINEERING")
print("=" * 95)


engineered_features = []


# ------------------------------------------------
# Salary gap
# ------------------------------------------------

if (
    "CurrentSalaryLKR" in X.columns
    and
    "ExpectedSalaryLKR" in X.columns
):

    X[
        "ExpectedSalaryGapLKR"
    ] = (

        X["ExpectedSalaryLKR"]

        -

        X["CurrentSalaryLKR"]

    )

    engineered_features.append(
        "ExpectedSalaryGapLKR"
    )


# ------------------------------------------------
# Salary ratio
# ------------------------------------------------

if (
    "CurrentSalaryLKR" in X.columns
    and
    "ExpectedSalaryLKR" in X.columns
):

    X[
        "ExpectedCurrentSalaryRatio"
    ] = (

        X["ExpectedSalaryLKR"]

        /

        X[
            "CurrentSalaryLKR"
        ].replace(
            0,
            np.nan
        )

    )

    engineered_features.append(
        "ExpectedCurrentSalaryRatio"
    )


# ------------------------------------------------
# Expected salary increase percentage
# ------------------------------------------------

if (
    "ExpectedSalaryGapLKR" in X.columns
    and
    "CurrentSalaryLKR" in X.columns
):

    X[
        "ExpectedSalaryIncreasePct"
    ] = (

        X[
            "ExpectedSalaryGapLKR"
        ]

        /

        X[
            "CurrentSalaryLKR"
        ].replace(
            0,
            np.nan
        )

        *

        100

    )

    engineered_features.append(
        "ExpectedSalaryIncreasePct"
    )


# ------------------------------------------------
# Satisfaction average
# ------------------------------------------------

SATISFACTION_COLUMNS = [

    column

    for column in [

        "EnvironmentSatisfaction",

        "JobSatisfaction",

        "RelationshipSatisfaction",

        "WorkLifeBalance",

        "JobInvolvement"

    ]

    if column in X.columns
]


if SATISFACTION_COLUMNS:

    X[
        "AverageSatisfaction"
    ] = (

        X[
            SATISFACTION_COLUMNS
        ].mean(
            axis=1
        )

    )

    X[
        "MinimumSatisfaction"
    ] = (

        X[
            SATISFACTION_COLUMNS
        ].min(
            axis=1
        )

    )

    X[
        "LowSatisfactionCount"
    ] = (

        X[
            SATISFACTION_COLUMNS
        ]
        .le(2)
        .sum(
            axis=1
        )

    )

    engineered_features += [

        "AverageSatisfaction",

        "MinimumSatisfaction",

        "LowSatisfactionCount"

    ]


# ------------------------------------------------
# Role tenure ratio
# ------------------------------------------------

if (
    "YearsInCurrentRole" in X.columns
    and
    "YearsAtCompany" in X.columns
):

    X[
        "RoleTenureRatio"
    ] = (

        X[
            "YearsInCurrentRole"
        ]

        /

        (
            X[
                "YearsAtCompany"
            ]

            +

            1
        )

    )

    engineered_features.append(
        "RoleTenureRatio"
    )


# ------------------------------------------------
# Manager tenure ratio
# ------------------------------------------------

if (
    "YearsWithCurrManager" in X.columns
    and
    "YearsAtCompany" in X.columns
):

    X[
        "ManagerTenureRatio"
    ] = (

        X[
            "YearsWithCurrManager"
        ]

        /

        (
            X[
                "YearsAtCompany"
            ]

            +

            1
        )

    )

    engineered_features.append(
        "ManagerTenureRatio"
    )


# ------------------------------------------------
# Promotion delay ratio
# ------------------------------------------------

if (
    "YearsSinceLastPromotion" in X.columns
    and
    "YearsAtCompany" in X.columns
):

    X[
        "PromotionDelayRatio"
    ] = (

        X[
            "YearsSinceLastPromotion"
        ]

        /

        (
            X[
                "YearsAtCompany"
            ]

            +

            1
        )

    )

    engineered_features.append(
        "PromotionDelayRatio"
    )


# ------------------------------------------------
# Company mobility rate
# ------------------------------------------------

if (
    "NumCompaniesWorked" in X.columns
    and
    "YearsAtCompany" in X.columns
):

    X[
        "CompanyMobilityRate"
    ] = (

        X[
            "NumCompaniesWorked"
        ]

        /

        (
            X[
                "YearsAtCompany"
            ]

            +

            1
        )

    )

    engineered_features.append(
        "CompanyMobilityRate"
    )


print(
    "Engineered Features:",
    len(engineered_features)
)

for feature in engineered_features:

    print(
        "•",
        feature
    )


# ================================================================
# 17. REMOVE CONSTANT FEATURES
# ================================================================

constant_features = [

    column

    for column in X.columns

    if X[
        column
    ].nunique(
        dropna=False
    ) <= 1
]


if constant_features:

    print(
        "\n⚠️ Constant features removed:"
    )

    for column in constant_features:

        print(
            "•",
            column
        )

    X = X.drop(
        columns=constant_features
    )


# ================================================================
# 18. DETECT DATA TYPES
# ================================================================

categorical_features = (

    X
    .select_dtypes(
        include=[
            "object",
            "category",
            "bool"
        ]
    )
    .columns
    .tolist()

)


numerical_features = [

    column

    for column in X.columns

    if column not in categorical_features

]


MODEL_FEATURES = (
    X.columns.tolist()
)


print("\n" + "=" * 95)
print("🔤 FEATURE TYPES")
print("=" * 95)

print(
    "Total Features:",
    len(MODEL_FEATURES)
)

print(
    "Categorical   :",
    len(categorical_features)
)

print(
    "Numerical     :",
    len(numerical_features)
)


# ================================================================
# 19. CLEAN CATEGORICAL DATA
# ================================================================

for column in categorical_features:

    X[column] = (

        X[column]

        .fillna(
            "Missing"
        )

        .astype(str)

        .str.strip()

    )


# ================================================================
# 20. CLEAN NUMERICAL DATA
# ================================================================

for column in numerical_features:

    X[column] = pd.to_numeric(
        X[column],
        errors="coerce"
    )

    X[column] = X[column].replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )

    median_value = (
        X[column].median()
    )

    if pd.isna(
        median_value
    ):

        median_value = 0

    X[column] = (
        X[column]
        .fillna(
            median_value
        )
    )


# ================================================================
# 21. FINAL DATASET SUMMARY
# ================================================================

print("\n" + "=" * 95)
print("📊 FINAL MODEL DATASET")
print("=" * 95)

print(
    "Samples        :",
    len(X)
)

print(
    "Features       :",
    len(X.columns)
)

print(
    "Numerical      :",
    len(numerical_features)
)

print(
    "Categorical    :",
    len(categorical_features)
)

print(
    "Attrition = 0  :",
    int((y == 0).sum())
)

print(
    "Attrition = 1  :",
    int((y == 1).sum())
)

print(
    "Missing Values :",
    int(X.isna().sum().sum())
)


# ================================================================
# 22. TRAIN / TEST SPLIT
# ================================================================

print("\n" + "=" * 95)
print("✂️ TRAIN / TEST SPLIT")
print("=" * 95)

X_train, X_test, y_train, y_test = (

    train_test_split(

        X,

        y,

        test_size=TEST_SIZE,

        stratify=y,

        random_state=RANDOM_STATE

    )

)


print(
    "Training Samples:",
    len(X_train)
)

print(
    "Testing Samples :",
    len(X_test)
)


# ================================================================
# 23. CROSS VALIDATION
# ================================================================

cv = StratifiedKFold(

    n_splits=N_FOLDS,

    shuffle=True,

    random_state=RANDOM_STATE

)


# ================================================================
# 24. LOGISTIC REGRESSION PREPROCESSOR
# ================================================================

logistic_preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",

            Pipeline(
                steps=[

                    (
                        "imputer",

                        SimpleImputer(
                            strategy="median"
                        )

                    ),

                    (
                        "scaler",

                        StandardScaler()
                    )

                ]
            ),

            numerical_features

        ),

        (
            "categorical",

            Pipeline(
                steps=[

                    (
                        "imputer",

                        SimpleImputer(
                            strategy="most_frequent"
                        )

                    ),

                    (
                        "onehot",

                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=True
                        )

                    )

                ]
            ),

            categorical_features

        )

    ],

    remainder="drop"

)


# ================================================================
# 25. LOGISTIC BASE MODEL
# ================================================================

logistic_model = Pipeline(

    steps=[

        (
            "preprocessor",
            logistic_preprocessor
        ),

        (
            "classifier",

            LogisticRegression(

                C=1.0,

                max_iter=2000,

                class_weight="balanced",

                solver="liblinear",

                random_state=RANDOM_STATE

            )

        )

    ]

)


# ================================================================
# 26. THRESHOLD FUNCTION
# ================================================================

def find_best_threshold(
    y_true,
    probabilities,
    minimum_recall=0.50
):

    best = None

    thresholds = np.arange(
        0.10,
        0.801,
        0.005
    )


    for threshold in thresholds:

        predictions = (

            probabilities >= threshold

        ).astype(int)


        recall = recall_score(

            y_true,

            predictions,

            zero_division=0

        )


        if recall < minimum_recall:

            continue


        accuracy = accuracy_score(

            y_true,

            predictions

        )


        precision = precision_score(

            y_true,

            predictions,

            zero_division=0

        )


        f1 = f1_score(

            y_true,

            predictions,

            zero_division=0

        )


        balanced = balanced_accuracy_score(

            y_true,

            predictions

        )


        candidate = {

            "threshold":
                float(threshold),

            "accuracy":
                float(accuracy),

            "precision":
                float(precision),

            "recall":
                float(recall),

            "f1":
                float(f1),

            "balanced_accuracy":
                float(balanced)

        }


        if best is None:

            best = candidate

        else:

            candidate_score = (

                candidate["accuracy"],

                candidate["f1"],

                candidate["balanced_accuracy"]

            )


            best_score = (

                best["accuracy"],

                best["f1"],

                best["balanced_accuracy"]

            )


            if candidate_score > best_score:

                best = candidate


    return best


# ================================================================
# 27. OPTUNA CATBOOST OBJECTIVE
# ================================================================

print("\n" + "=" * 95)
print("🔎 OPTUNA HYPERPARAMETER OPTIMIZATION")
print("=" * 95)

print(
    "Trials:",
    N_TRIALS
)

print(
    "CV Folds:",
    N_FOLDS
)

print(
    "Approximate CatBoost Fits:",
    N_TRIALS * N_FOLDS
)


def catboost_objective(
    trial
):

    iterations = trial.suggest_int(

        "iterations",

        300,

        900,

        step=100

    )


    depth = trial.suggest_int(

        "depth",

        4,

        7

    )


    learning_rate = trial.suggest_float(

        "learning_rate",

        0.01,

        0.08,

        log=True

    )


    l2_leaf_reg = trial.suggest_float(

        "l2_leaf_reg",

        2.0,

        30.0,

        log=True

    )


    random_strength = trial.suggest_float(

        "random_strength",

        0.0,

        4.0

    )


    bagging_temperature = trial.suggest_float(

        "bagging_temperature",

        0.0,

        4.0

    )


    border_count = trial.suggest_categorical(

        "border_count",

        [
            64,
            128,
            254
        ]

    )


    positive_class_weight = (

        trial.suggest_float(

            "positive_class_weight",

            1.0,

            3.0

        )

    )


    oof_probabilities = np.zeros(

        len(X_train),

        dtype=float

    )


    for fold, (
        train_indices,
        validation_indices
    ) in enumerate(

        cv.split(
            X_train,
            y_train
        ),

        start=1

    ):


        X_fold_train = (

            X_train
            .iloc[
                train_indices
            ]
        )


        y_fold_train = (

            y_train
            .iloc[
                train_indices
            ]
        )


        X_fold_validation = (

            X_train
            .iloc[
                validation_indices
            ]
        )


        model = CatBoostClassifier(

            iterations=iterations,

            depth=depth,

            learning_rate=learning_rate,

            l2_leaf_reg=l2_leaf_reg,

            random_strength=random_strength,

            bagging_temperature=bagging_temperature,

            border_count=border_count,

            boosting_type="Ordered",

            class_weights=[
                1.0,
                positive_class_weight
            ],

            loss_function="Logloss",

            eval_metric="AUC",

            random_seed=(
                RANDOM_STATE + fold
            ),

            verbose=False,

            allow_writing_files=False,

            thread_count=-1

        )


        model.fit(

            X_fold_train,

            y_fold_train,

            cat_features=categorical_features

        )


        oof_probabilities[
            validation_indices
        ] = (

            model
            .predict_proba(
                X_fold_validation
            )[:, 1]

        )


    threshold_result = find_best_threshold(

        y_train,

        oof_probabilities,

        minimum_recall=MIN_ATTRITION_RECALL

    )


    if threshold_result is None:

        return 0.0


    trial.set_user_attr(

        "threshold",

        threshold_result["threshold"]

    )


    trial.set_user_attr(

        "accuracy",

        threshold_result["accuracy"]

    )


    trial.set_user_attr(

        "precision",

        threshold_result["precision"]

    )


    trial.set_user_attr(

        "recall",

        threshold_result["recall"]

    )


    trial.set_user_attr(

        "f1",

        threshold_result["f1"]

    )


    trial.set_user_attr(

        "balanced_accuracy",

        threshold_result[
            "balanced_accuracy"
        ]

    )


    trial.set_user_attr(

        "roc_auc",

        roc_auc_score(

            y_train,

            oof_probabilities

        )

    )


    trial.set_user_attr(

        "pr_auc",

        average_precision_score(

            y_train,

            oof_probabilities

        )

    )


    # Accuracy is primary objective.
    # F1 and balanced accuracy provide tiny tie-break bonuses.

    score = (

        threshold_result[
            "accuracy"
        ]

        +

        0.001

        *

        threshold_result[
            "f1"
        ]

        +

        0.0001

        *

        threshold_result[
            "balanced_accuracy"
        ]

    )


    return score


# ================================================================
# 28. RUN OPTUNA
# ================================================================

sampler = optuna.samplers.TPESampler(

    seed=RANDOM_STATE

)


study = optuna.create_study(

    direction="maximize",

    sampler=sampler

)


study.optimize(

    catboost_objective,

    n_trials=N_TRIALS,

    show_progress_bar=True,

    gc_after_trial=True

)


# ================================================================
# 29. BEST OPTUNA RESULT
# ================================================================

best_trial = (
    study.best_trial
)


best_params = (
    best_trial.params.copy()
)


best_threshold = float(

    best_trial.user_attrs[
        "threshold"
    ]

)


positive_class_weight = (

    best_params.pop(
        "positive_class_weight"
    )

)


print("\n" + "=" * 95)
print("🏆 BEST CATBOOST PARAMETERS")
print("=" * 95)

for key, value in best_params.items():

    print(
        f"{key:25s}: {value}"
    )


print(
    "\nPositive Class Weight:",
    round(
        positive_class_weight,
        4
    )
)


print(
    "\nOOF Threshold:",
    f"{best_threshold:.4f}"
)


print(
    "OOF Accuracy:",
    f"{best_trial.user_attrs['accuracy']:.4f}"
)


print(
    "OOF Precision:",
    f"{best_trial.user_attrs['precision']:.4f}"
)


print(
    "OOF Recall:",
    f"{best_trial.user_attrs['recall']:.4f}"
)


print(
    "OOF F1:",
    f"{best_trial.user_attrs['f1']:.4f}"
)


print(
    "OOF ROC-AUC:",
    f"{best_trial.user_attrs['roc_auc']:.4f}"
)


print(
    "OOF PR-AUC:",
    f"{best_trial.user_attrs['pr_auc']:.4f}"
)


# ================================================================
# 30. TRAIN FINAL CATBOOST
# ================================================================

print("\n" + "=" * 95)
print("🧠 TRAINING FINAL BASE MODELS")
print("=" * 95)


final_catboost = CatBoostClassifier(

    **best_params,

    class_weights=[
        1.0,
        positive_class_weight
    ],

    boosting_type="Ordered",

    loss_function="Logloss",

    eval_metric="AUC",

    random_seed=RANDOM_STATE,

    verbose=False,

    allow_writing_files=False,

    thread_count=-1

)


final_catboost.fit(

    X_train,

    y_train,

    cat_features=categorical_features

)


print(
    "✅ CatBoost trained"
)


# ================================================================
# 31. TRAIN LOGISTIC REGRESSION
# ================================================================

final_logistic = Pipeline(

    steps=[

        (
            "preprocessor",

            ColumnTransformer(

                transformers=[

                    (
                        "numeric",

                        Pipeline(
                            steps=[

                                (
                                    "imputer",

                                    SimpleImputer(
                                        strategy="median"
                                    )

                                ),

                                (
                                    "scaler",

                                    StandardScaler()
                                )

                            ]
                        ),

                        numerical_features

                    ),

                    (
                        "categorical",

                        Pipeline(
                            steps=[

                                (
                                    "imputer",

                                    SimpleImputer(
                                        strategy="most_frequent"
                                    )

                                ),

                                (
                                    "onehot",

                                    OneHotEncoder(
                                        handle_unknown="ignore",
                                        sparse_output=True
                                    )

                                )

                            ]
                        ),

                        categorical_features

                    )

                ]

            )

        ),

        (
            "classifier",

            LogisticRegression(

                C=1.0,

                max_iter=2000,

                class_weight="balanced",

                solver="liblinear",

                random_state=RANDOM_STATE

            )

        )

    ]

)


final_logistic.fit(

    X_train,

    y_train

)


print(
    "✅ Logistic Regression trained"
)


# ================================================================
# 32. CREATE OOF STACKING DATA
# ================================================================

print("\n" + "=" * 95)
print("🔗 BUILDING HYBRID STACKING DATA")
print("=" * 95)


oof_catboost = np.zeros(
    len(X_train)
)


oof_logistic = np.zeros(
    len(X_train)
)


for fold, (
    train_indices,
    validation_indices
) in enumerate(

    cv.split(
        X_train,
        y_train
    ),

    start=1

):

    print(
        f"Fold {fold}/{N_FOLDS}"
    )


    X_fold_train = (
        X_train
        .iloc[
            train_indices
        ]
    )


    y_fold_train = (
        y_train
        .iloc[
            train_indices
        ]
    )


    X_fold_validation = (
        X_train
        .iloc[
            validation_indices
        ]
    )


    # ------------------------------------------------------------
    # CatBoost
    # ------------------------------------------------------------

    fold_catboost = CatBoostClassifier(

        **best_params,

        class_weights=[
            1.0,
            positive_class_weight
        ],

        boosting_type="Ordered",

        loss_function="Logloss",

        eval_metric="AUC",

        random_seed=(
            RANDOM_STATE + fold
        ),

        verbose=False,

        allow_writing_files=False,

        thread_count=-1

    )


    fold_catboost.fit(

        X_fold_train,

        y_fold_train,

        cat_features=categorical_features

    )


    oof_catboost[
        validation_indices
    ] = (

        fold_catboost
        .predict_proba(
            X_fold_validation
        )[:, 1]

    )


    # ------------------------------------------------------------
    # Logistic Regression
    # ------------------------------------------------------------

    fold_logistic = clone(
        logistic_model
    )


    fold_logistic.fit(

        X_fold_train,

        y_fold_train

    )


    oof_logistic[
        validation_indices
    ] = (

        fold_logistic
        .predict_proba(
            X_fold_validation
        )[:, 1]

    )


# ================================================================
# 33. META-LEARNER DATA
# ================================================================

meta_X_train = pd.DataFrame({

    "CatBoostProbability":
        oof_catboost,

    "LogisticProbability":
        oof_logistic

})


print(
    "\nMeta features:"
)

display(
    meta_X_train.head()
)


# ================================================================
# 34. TRAIN META LEARNER
# ================================================================

meta_model = LogisticRegression(

    C=1.0,

    max_iter=1000,

    class_weight="balanced",

    random_state=RANDOM_STATE

)


meta_model.fit(

    meta_X_train,

    y_train

)


print(
    "✅ Meta learner trained"
)


# ================================================================
# 35. OOF STACKING PROBABILITIES
# ================================================================

oof_stacking_probability = (

    meta_model
    .predict_proba(
        meta_X_train
    )[:, 1]

)


# ================================================================
# 36. FINAL STACKING THRESHOLD
# ================================================================

stacking_threshold_result = (

    find_best_threshold(

        y_train,

        oof_stacking_probability,

        minimum_recall=MIN_ATTRITION_RECALL

    )

)


if stacking_threshold_result is None:

    raise ValueError(
        "❌ Unable to find a valid threshold."
    )


stacking_threshold = (

    stacking_threshold_result[
        "threshold"
    ]

)


print("\n" + "=" * 95)
print("🎯 STACKING THRESHOLD")
print("=" * 95)

print(
    "Threshold:",
    f"{stacking_threshold:.4f}"
)

print(
    "OOF Accuracy:",
    f"{stacking_threshold_result['accuracy']:.4f}"
)

print(
    "OOF Precision:",
    f"{stacking_threshold_result['precision']:.4f}"
)

print(
    "OOF Recall:",
    f"{stacking_threshold_result['recall']:.4f}"
)

print(
    "OOF F1:",
    f"{stacking_threshold_result['f1']:.4f}"
)


# ================================================================
# 37. TEST SET BASE MODEL PREDICTIONS
# ================================================================

test_catboost_probability = (

    final_catboost
    .predict_proba(
        X_test
    )[:, 1]

)


test_logistic_probability = (

    final_logistic
    .predict_proba(
        X_test
    )[:, 1]

)


# ================================================================
# 38. TEST META MODEL
# ================================================================

meta_X_test = pd.DataFrame({

    "CatBoostProbability":
        test_catboost_probability,

    "LogisticProbability":
        test_logistic_probability

})


test_stacking_probability = (

    meta_model
    .predict_proba(
        meta_X_test
    )[:, 1]

)


# ================================================================
# 39. FINAL TEST PREDICTIONS
# ================================================================

test_predictions = (

    test_stacking_probability
    >= stacking_threshold

).astype(int)


# ================================================================
# 40. FINAL METRICS
# ================================================================

accuracy = accuracy_score(

    y_test,

    test_predictions

)


balanced_accuracy = balanced_accuracy_score(

    y_test,

    test_predictions

)


precision = precision_score(

    y_test,

    test_predictions,

    zero_division=0

)


recall = recall_score(

    y_test,

    test_predictions,

    zero_division=0

)


f1 = f1_score(

    y_test,

    test_predictions,

    zero_division=0

)


roc_auc = roc_auc_score(

    y_test,

    test_stacking_probability

)


pr_auc = average_precision_score(

    y_test,

    test_stacking_probability

)


# ================================================================
# 41. FINAL RESULTS
# ================================================================

print("\n" + "=" * 95)
print("🏁 FINAL HYBRID STACKING RESULTS")
print("=" * 95)

print(
    f"Majority Baseline : {majority_baseline * 100:.2f}%"
)

print(
    f"Accuracy          : {accuracy * 100:.2f}%"
)

print(
    f"Balanced Accuracy : {balanced_accuracy * 100:.2f}%"
)

print(
    f"Precision         : {precision * 100:.2f}%"
)

print(
    f"Recall            : {recall * 100:.2f}%"
)

print(
    f"F1 Score          : {f1:.4f}"
)

print(
    f"ROC-AUC           : {roc_auc:.4f}"
)

print(
    f"PR-AUC            : {pr_auc:.4f}"
)

print(
    f"Threshold         : {stacking_threshold:.4f}"
)


# ================================================================
# 42. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(

    y_test,

    test_predictions

)


tn, fp, fn, tp = cm.ravel()


print("\n" + "=" * 95)
print("📌 CONFUSION MATRIX")
print("=" * 95)

print(
    cm
)

print(
    "\nTrue Negative :",
    tn
)

print(
    "False Positive:",
    fp
)

print(
    "False Negative:",
    fn
)

print(
    "True Positive :",
    tp
)


# ================================================================
# 43. CONFUSION MATRIX VISUALIZATION
# ================================================================

plt.figure(figsize=(7, 6))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Hybrid Stacking Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    [0, 1],
    ["No Attrition", "Attrition"]
)

plt.yticks(
    [0, 1],
    ["No Attrition", "Attrition"]
)


for i in range(2):

    for j in range(2):

        plt.text(

            j,

            i,

            cm[i, j],

            ha="center",

            va="center"

        )


plt.colorbar()

plt.tight_layout()

plt.show()


# ================================================================
# 44. CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 95)
print("📋 CLASSIFICATION REPORT")
print("=" * 95)

print(

    classification_report(

        y_test,

        test_predictions,

        target_names=[
            "No Attrition",
            "Attrition"
        ],

        digits=4,

        zero_division=0

    )

)


# ================================================================
# 48. MODEL COMPARISON
# ================================================================

catboost_test_prediction = (

    test_catboost_probability
    >= stacking_threshold

).astype(int)


logistic_test_prediction = (

    test_logistic_probability
    >= stacking_threshold

).astype(int)


model_comparison = pd.DataFrame({

    "Model": [

        "CatBoost",

        "Logistic Regression",

        "Hybrid Stacking"

    ],

    "Accuracy": [

        accuracy_score(
            y_test,
            catboost_test_prediction
        ),

        accuracy_score(
            y_test,
            logistic_test_prediction
        ),

        accuracy

    ],

    "Precision": [

        precision_score(
            y_test,
            catboost_test_prediction,
            zero_division=0
        ),

        precision_score(
            y_test,
            logistic_test_prediction,
            zero_division=0
        ),

        precision

    ],

    "Recall": [

        recall_score(
            y_test,
            catboost_test_prediction,
            zero_division=0
        ),

        recall_score(
            y_test,
            logistic_test_prediction,
            zero_division=0
        ),

        recall

    ],

    "F1": [

        f1_score(
            y_test,
            catboost_test_prediction,
            zero_division=0
        ),

        f1_score(
            y_test,
            logistic_test_prediction,
            zero_division=0
        ),

        f1

    ],

    "ROC-AUC": [

        roc_auc_score(
            y_test,
            test_catboost_probability
        ),

        roc_auc_score(
            y_test,
            test_logistic_probability
        ),

        roc_auc

    ],

    "PR-AUC": [

        average_precision_score(
            y_test,
            test_catboost_probability
        ),

        average_precision_score(
            y_test,
            test_logistic_probability
        ),

        pr_auc

    ]

})


print("\n" + "=" * 95)
print("📊 BASE MODEL VS HYBRID STACKING")
print("=" * 95)

display(
    model_comparison.round(4)
)


# ================================================================
# 50. CATBOOST FEATURE IMPORTANCE
# ================================================================

importance_df = pd.DataFrame({

    "Feature":
        final_catboost.feature_names_,

    "Importance":
        final_catboost.get_feature_importance()

})


importance_df = (

    importance_df

    .sort_values(

        by="Importance",

        ascending=False

    )

    .reset_index(
        drop=True
    )

)


print("\n" + "=" * 95)
print("🔎 TOP 20 CATBOOST FEATURES")
print("=" * 95)

display(

    importance_df
    .head(20)
    .round(4)

)


# ================================================================
# 51. FEATURE IMPORTANCE VISUALIZATION
# ================================================================

top_features = (

    importance_df
    .head(15)
    .sort_values(
        "Importance"
    )

)


plt.figure(figsize=(9, 7))

plt.barh(

    top_features["Feature"],

    top_features["Importance"]

)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 15 CatBoost Feature Importances"
)

plt.tight_layout()

plt.show()


# ================================================================
# 52. STACKING META-LEARNER COEFFICIENTS
# ================================================================

meta_coefficients = pd.DataFrame({

    "Base Model": [

        "CatBoost",

        "Logistic Regression"

    ],

    "Coefficient": (

        meta_model
        .coef_[0]

    )

})


print("\n" + "=" * 95)
print("🔗 META-LEARNER COEFFICIENTS")
print("=" * 95)

display(
    meta_coefficients.round(4)
)


# ================================================================
# 53. SAMPLE PREDICTION
# ================================================================

sample = (

    X_test
    .iloc[[0]]
    .copy()

)


sample_catboost_probability = (

    final_catboost
    .predict_proba(
        sample
    )[0][1]

)


sample_logistic_probability = (

    final_logistic
    .predict_proba(
        sample
    )[0][1]

)


sample_meta = pd.DataFrame({

    "CatBoostProbability": [

        sample_catboost_probability

    ],

    "LogisticProbability": [

        sample_logistic_probability

    ]

})


sample_probability = (

    meta_model
    .predict_proba(
        sample_meta
    )[0][1]

)


sample_prediction = int(

    sample_probability
    >= stacking_threshold

)


def get_risk_level(
    probability
):

    percentage = (
        probability * 100
    )

    if percentage < 30:

        return "LOW"

    elif percentage < 60:

        return "MEDIUM"

    else:

        return "HIGH"


print("\n" + "=" * 95)
print("🧪 SAMPLE PREDICTION")
print("=" * 95)

print(
    f"Attrition Probability : "
    f"{sample_probability * 100:.2f}%"
)

print(
    "Predicted Class       :",
    sample_prediction
)

print(
    "Risk Level            :",
    get_risk_level(
        sample_probability
    )
)


# ================================================================
# 54. CATEGORY VALUES
# ================================================================

category_values = {}


for column in categorical_features:

    category_values[column] = (

        X[column]

        .dropna()

        .astype(str)

        .unique()

        .tolist()

    )


# ================================================================
# 55. MODEL PACKAGE
# ================================================================

model_package = {

    "model":
        final_catboost,

    "logistic_model":
        final_logistic,

    "meta_model":
        meta_model,

    "threshold":
        float(
            stacking_threshold
        ),

    "features":
        FEATURES,

    "model_features":
        MODEL_FEATURES,

    "categorical_features":
        categorical_features,

    "numerical_features":
        numerical_features,

    "category_values":
        category_values,

    "target":
        TARGET,

    "target_source":
        TARGET_SOURCE,

    "model_name":
        "Sri Lankan Employee Attrition V10 Hybrid Stacking",

    "base_models": [

        "CatBoost",

        "LogisticRegression"

    ],

    "meta_model_name":
        "LogisticRegression",

    "best_catboost_parameters":
        best_params,

    "positive_class_weight":
        float(
            positive_class_weight
        ),

    "cv_settings": {

        "folds":
            N_FOLDS,

        "test_size":
            TEST_SIZE,

        "minimum_attrition_recall":
            MIN_ATTRITION_RECALL

    },

    "excluded_columns":
        DROP_COLUMNS,

    "engineered_features":
        engineered_features,

    "metrics": {

        "majority_baseline":
            float(
                majority_baseline
            ),

        "accuracy":
            float(
                accuracy
            ),

        "balanced_accuracy":
            float(
                balanced_accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1":
            float(
                f1
            ),

        "roc_auc":
            float(
                roc_auc
            ),

        "pr_auc":
            float(
                pr_auc
            )

    }

}


# ================================================================
# 56. SAVE JOBLIB MODEL
# ================================================================

MODEL_FILE = (

    "sri_lankan_attrition_v10_hybrid_stacking.joblib"

)


joblib.dump(

    model_package,

    MODEL_FILE

)


print("\n" + "=" * 95)
print("💾 MODEL SAVED")
print("=" * 95)

print(
    "✅",
    MODEL_FILE
)


# ================================================================
# 57. SAVE CATBOOST MODEL
# ================================================================

CATBOOST_FILE = (

    "sri_lankan_attrition_v10_catboost.cbm"

)


final_catboost.save_model(

    CATBOOST_FILE

)


print(
    "✅",
    CATBOOST_FILE
)


# ================================================================
# 58. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 95)
print("🏁 FINAL MODEL SUMMARY")
print("=" * 95)

print(
    f"Dataset Rows       : {len(df)}"
)

print(
    f"Model Features     : {len(MODEL_FEATURES)}"
)

print(
    f"Training Samples   : {len(X_train)}"
)

print(
    f"Testing Samples    : {len(X_test)}"
)

print(
    f"Majority Baseline  : {majority_baseline * 100:.2f}%"
)

print(
    f"Model Accuracy     : {accuracy * 100:.2f}%"
)

print(
    f"Balanced Accuracy  : {balanced_accuracy * 100:.2f}%"
)

print(
    f"Attrition Precision: {precision * 100:.2f}%"
)

print(
    f"Attrition Recall   : {recall * 100:.2f}%"
)

print(
    f"F1 Score           : {f1:.4f}"
)

print(
    f"ROC-AUC            : {roc_auc:.4f}"
)

print(
    f"PR-AUC             : {pr_auc:.4f}"
)

print(
    f"Threshold          : {stacking_threshold:.4f}"
)

print(
    "\nArchitecture:"
)

print(
    "CatBoost →"
    " Logistic Regression →"
    " Meta Learner"
)

print(
    "\n💾 Saved Files:"
)

print(
    "•",
    MODEL_FILE
)

print(
    "•",
    CATBOOST_FILE
)


# ================================================================
# 59. DOWNLOAD FINAL MODEL PACKAGE
# ================================================================

print(
    "\n📥 Downloading model package..."
)

files.download(
    MODEL_FILE
)

print(
    "\n✅ V10 TRAINING COMPLETED SUCCESSFULLY"
)

print("=" * 95)
```
